1. Write a recursive function power(base, exponent) that, when called, returns base^exponent<br>
Example: power(3,4) = 3\*3\*3\*3. Both base and exponent are user-defined and assume that the
exponent is an integer greater than or equal to 1.

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
def power(base, exponent):
    if exponent == 0:
        return 1
    return base * power(base, exponent - 1)
print(power(3, 4))
```

2. The greatest common divisor of integers x and y is the largest integer that evenly divides into both x
and y. Write and test a recursive function gcd that returns the greatest common divisor of x and y.
The gcd of x and y is defined recursively as follows: If y is equal to 0, then gcd(x, y) is x; otherwise,
gcd(x, y) is gcd(y, x%y).

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
def gcd(x, y):
    if y == 0:
        return x
    return gcd(y, x % y)
print(gcd(48, 18)) 
```

4. Implement a recursive solution for computing the nth Fibonacci number. Then, analyze its time
complexity. Propose a more efficient solution and compare the two approaches.

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
# Time Complexity: 𝑂(2^𝑛) (exponential)
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
print(fibonacci(6)) 

# Time Complexity: 𝑂(𝑛) (linear)
def fibonacci_optimized(n):
    fib = [0, 1]
    for i in range(2, n + 1):
        fib.append(fib[i - 1] + fib[i - 2])
    return fib[n]
print(fibonacci_optimized(6))
```

5. Given an array of N elements, not necessarily in ascending order, devised an algorithm to find the
kth largest one. It should run in O(N) time on random inputs.

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
# Time Complexity: 𝑂(𝑁) on average
import random

def partition(arr, left, right, pivot_index):
    pivot = arr[pivot_index]
    arr[pivot_index], arr[right] = arr[right], arr[pivot_index]
    store_index = left
    for i in range(left, right):
        if arr[i] > pivot: 
            arr[i], arr[store_index] = arr[store_index], arr[i]
            store_index += 1
    arr[store_index], arr[right] = arr[right], arr[store_index]
    return store_index

def quickselect(arr, left, right, k):
    if left == right:
        return arr[left]
    pivot_index = random.randint(left, right)
    pivot_index = partition(arr, left, right, pivot_index)
    if k == pivot_index:
        return arr[k]
    elif k < pivot_index:
        return quickselect(arr, left, pivot_index - 1, k)
    else:
        return quickselect(arr, pivot_index + 1, right, k)

def kth_largest(arr, k):
    return quickselect(arr, 0, len(arr) - 1, k - 1)

arr = [3, 2, 1, 5, 6, 4]
print(kth_largest(arr, 2)) 
```

7. Given N points on a circle, centered at the origin, design an algorithm that determines whether there
are two points that are antipodal, i.e., the line connecting the two points goes through the origin. Your
algorithm should run in time proportional to NlogN.

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
# Time Complexity: 𝑂(𝑁log⁡𝑁) (due to sorting)
def find_antipodal_points(points):
    points.sort()
    n = len(points)
    for i in range(n):
        if points[i] + points[(i + n // 2) % n] == 0:
            return points[i], points[(i + n // 2) % n]
    return None
points = [-3, -1, 1, 3]
print(find_antipodal_points(points)) 
```

8. Write a Python function quick sort that implements the quicksort algorithm. The function should
include a helper function quick sort helper to handle recursion. The helper function must take
a starting and ending index as arguments and sort the array in-place. Demonstrate the function by
sorting the given array and printing the sorted output.

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
# Time Complexity: 𝑂(𝑛log⁡𝑛) (average case)
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[0]
    left = [x for x in arr[1:] if x <= pivot]
    right = [x for x in arr[1:] if x > pivot]
    return quick_sort(left) + [pivot] + quick_sort(right)
arr = [37, 2, 6, 4, 89, 8, 10, 12, 68, 45]
print(quick_sort(arr))
```

9. You are given the following list of famous personalities with their net worth:<br>
• Elon Musk: $433.9 Billion<br>
• Jeff Bezos: $239.4 Billion<br>
• Mark Zuckerberg: $211.8 Billion<br>
• Larry Ellison: $204.6 Billion<br>
• Bernard Arnault & Family: $181.3 Billion<br>
• Larry Page: $161.4 Billion<br>
Develop a program to sort the aforementioned details on the basis of net worth using<br>
a. Selection sort<br>
b. Bubble sort<br>
c. Insertion sort.<br>
The final sorted data should be the same for all cases. After you obtain the sorted data, present the<br>
result in the form of the following dictionary:<br>
{’name1’:networth1, ’name2’:networth2,...}<br>

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
# Time Complexity: 𝑂(𝑛^2)for all three algorithms

# Selection Sort
def selection_sort(data):
    items = list(data.items())
    n = len(items)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if items[j][1] < items[min_idx][1]:
                min_idx = j
        items[i], items[min_idx] = items[min_idx], items[i]
    return dict(items)

# Bubble Sort
def bubble_sort(data):
    items = list(data.items())
    n = len(items)
    for i in range(n):
        for j in range(n - i - 1):
            if items[j][1] > items[j + 1][1]:
                items[j], items[j + 1] = items[j + 1], items[j]
    return dict(items)

# Insertion Sort
def insertion_sort(data):
    items = list(data.items())
    n = len(items)
    for i in range(1, n):
        key = items[i]
        j = i - 1
        while j >= 0 and key[1] < items[j][1]:
            items[j + 1] = items[j]
            j -= 1
        items[j + 1] = key
    return dict(items)

net_worths = {
    "Elon Musk": 433.9, "Jeff Bezos": 239.4, "Mark Zuckerberg": 211.8,
    "Larry Ellison": 204.6, "Bernard Arnault & Family": 181.3, "Larry Page": 161.4
}
print(selection_sort(net_worths))
print(bubble_sort(net_worths))
print(insertion_sort(net_worths))
```

10. Use Merge Sort to sort a list of strings alphabetically. Example:<br>
Input: [’apple’, ’orange’, ’banana’, ’grape’]<br>
Output: [’apple’, ’banana’, ’grape’, ’orange’]<br>

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    while left and right:
        if left[0] < right[0]:
            result.append(left.pop(0))
        else:
            result.append(right.pop(0))
    return result + left + right

strings = ['apple', 'orange', 'banana', 'grape']
print(merge_sort(strings))
```

11. Without using the built-in sorted() function, write a Python program to merge two pre-sorted lists into
a single sorted list using the logic of Merge Sort. Example:<br>
Input: [1, 3, 5, 7] and [2, 4, 6, 8]<br>
Output: [1, 2, 3, 4, 5, 6, 7, 8]<br>


```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
def merge_sorted_lists(list1, list2):
    result = []
    i, j = 0, 0
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1
    result.extend(list1[i:])
    result.extend(list2[j:])
    return result

list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]
print(merge_sorted_lists(list1, list2))
```
