1. Define Natural Language Processing (NLP). Provide three real-world applications of NLP and explain how they impact society.<br>
<br>
NLP Definition:<br>
Natural Language Processing (NLP) is a field of Artificial Intelligence that enables machines to understand, interpret, and respond to human language in a meaningful way. It combines linguistics and machine learning to process text or speech data.<br>
Applications and Impact:<br>
i) Chatbots and Virtual Assistants: Improve customer service by providing instant responses and reducing human intervention.<br>
ii) Sentiment Analysis: Analyzes customer feedback and social media data to identify opinions and sentiments.<br>
iii) Machine Translation (e.g., Google Translate): Breaks language barriers and enhances global communication.<br>

2. Explain the following terms and their significance in NLP:<br>
<br>
Tokenization: Splitting text into smaller units, such as words or sentences, which helps in text preprocessing.<br>
Stemming: Reducing words to their root form by removing suffixes, improving search relevance.<br>
Lemmatization: Converting words to their base or dictionary form, preserving meaning better than stemming.<br>

3. What is Part-of-Speech (POS) tagging? Discuss its importance with an example.<br>
<br>
POS Tagging:<br>
It assigns parts of speech (noun, verb, adjective, etc.) to words in a sentence. It is important in syntactic parsing, named entity recognition, and sentiment analysis.<br>
<br>
Example:<br>
Text: "The cat is sleeping."<br>
POS Tags:<br>
The (Determiner)<br>
cat (Noun)<br>
is (Verb)<br>
sleeping (Verb)<br>

5. Write a Python script to perform the following tasks on the given text:

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from textblob import TextBlob

text = """Natural Language Processing enables machines to understand and process human languages. 
It is a fascinating field with numerous applications, such as chatbots and language translation."""

blob = TextBlob(text)
words = blob.words
sentences = blob.sentences

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stemmed_words = [stemmer.stem(word) for word in words]
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words]

print(f"Words: {words}")
print(f"Stemmed Words: {stemmed_words}")
print(f"Lemmatized Words: {lemmatized_words}")
print(f"Filtered Words: {filtered_words}")
```

10. Spell Check a Project Gutenberg Book

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
from textblob import TextBlob
import requests

url = "https://www.gutenberg.org/files/1342/1342-0.txt"
response = requests.get(url)
book_text = response.text

blob = TextBlob(book_text)

misspelled = {}
for word in blob.words:
    if not word.correct() == word:
        misspelled[word] = word.spellcheck()

for word, suggestions in list(misspelled.items())[:5]:
    print(f"{word} -> {suggestions}")
```

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
# h) Summarize a paragraph based on noun phrase frequency
from textblob import TextBlob

text = input("Enter a paragraph: ")
blob = TextBlob(text)

sentence_scores = {}
for sentence in blob.sentences:
    score = sum(1 for phrase in blob.noun_phrases if phrase in sentence.lower())
    sentence_scores[sentence] = score

top_sentences = sorted(sentence_scores, key=sentence_scores.get, reverse=True)[:3]
summary = ' '.join(str(sentence) for sentence in top_sentences)
print("Summary:", summary)
```

12. Python Program to Get Word Definition, Synonyms, and Antonyms

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
from nltk.corpus import wordnet

word = input("Enter a word: ")

synsets = wordnet.synsets(word)
if synsets:
    print(f"Definition of '{word}': {synsets[0].definition()}")

    synonyms = set()
    antonyms = set()
    for syn in synsets:
        for lemma in syn.lemmas():
            synonyms.add(lemma.name())
            if lemma.antonyms():
                antonyms.add(lemma.antonyms()[0].name())

    print(f"Synonyms of '{word}': {list(synonyms)}")
    print(f"Antonyms of '{word}': {list(antonyms) if antonyms else 'None'}")
else:
    print(f"No definition found for '{word}'")
```

16. spaCy: Shakespeare Similarity Detection

```python
# Name: Aditya Prasad Sahu
# Regd. No.: 2241019166
import spacy
import requests

nlp = spacy.load('en_core_web_sm')

comedy_url = "https://www.gutenberg.org/files/2236/2236-0.txt"
romeo_url = "https://www.gutenberg.org/files/1513/1513-0.txt"

comedy_text = requests.get(comedy_url).text
romeo_text = requests.get(romeo_url).text

comedy_doc = nlp(comedy_text)
romeo_doc = nlp(romeo_text)

similarity_score = comedy_doc.similarity(romeo_doc)
print(f"Similarity Score: {similarity_score}")
```
