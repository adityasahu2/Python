```python
## 1. Sum of First n Natural Numbers (Recursion vs Iteration)

def recursive_sum(n):
    if n == 1:
        return 1
    return n + recursive_sum(n - 1)

def iterative_sum(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

# Time Complexity Analysis
# Recursive: O(n) (due to n recursive calls)
# Iterative: O(n) (due to single loop iteration)
# Iteration is more efficient due to no function call overhead.
```

```python
## 2. Josephus Problem

def josephus(n, k):
    if n == 1:
        return 0
    return (josephus(n - 1, k) + k) % n

# Base Case: When one person remains.
# Recursive Formula: (josephus(n - 1, k) + k) % n
# Time Complexity: O(n)
```

```python

## 3. Exponentiation by Squaring (Recursive and Iterative)

def fast_power(a, b):
    if b == 0:
        return 1
    half = fast_power(a, b // 2)
    if b % 2 == 0:
        return half * half
    else:
        return a * half * half

def fast_power_iterative(a, b):
    result = 1
    while b:
        if b % 2 == 1:
            result *= a
        a *= a
        b //= 2
    return result

# Recursive: O(log b)
# Iterative: O(log b), avoids function call overhead.

```

```python
## 4. Phantom Tracking - Identifying Suspicious Transactions

def linear_search(transactions, threshold):
    return [t for t in transactions if t > threshold]

transactions = [120, 45, 300, 220, 90, 600, 130, 75, 800, 500, 350, 40]
threshold = 250
suspicious = linear_search(transactions, threshold)

# Time Complexity: O(n)
```

```python

## 5. Sorting Suspicious Transactions using Selection Sort

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

sorted_suspicious = selection_sort(suspicious)

# Time Complexity: O(n^2)

```

```python
## 6. Binary Search for a Specific Transaction

def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return True
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return False

search_amount = 500
exists = binary_search(sorted_suspicious, search_amount)

# Time Complexity: O(log n)

```

```python
## 7. Reconstructing Full Transaction History with Merge Sort

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    sorted_arr = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_arr.append(left[i])
            i += 1
        else:
            sorted_arr.append(right[j])
            j += 1
    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])
    return sorted_arr

sorted_transactions = merge_sort(transactions)

# Time Complexity: O(n log n)
```
