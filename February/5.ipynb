```python
#__str__and__repr__
class Person:
    def __init__(self, name,age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f'Person: {self.name}, Age: {self.age}'
    
    def __repr__(self):
        return f'Person({self.name}, {self.age})'
person = Person("Aditya" ,20)
print(str(person))
print(repr(person))
```

```python
#@property (Decorator for Getter)
class Circle:
    def __init__(self, radius):
        self._radius = radius #Private attribute
    
    @property
    def radius(self):
        #Getter using @property decorator to return radius
        return self._radius
circle = Circle(5)
print(circle.radius)
```

```python
#setter method
class Circle:
    def __init__(self, radius):
        self._radius = radius #Private attribute
    
    @property
    def radius(self):
        #Getter using @property decorator to return radius
        return self._radius
    def set_radius(self, radius):
        #Setter method to set radius
        if radius < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = radius

circle = Circle(5)
print(circle.radius)
circle.set_radius(3)
print(circle.radius)
```

```python
# @property (decorator for setter)
class Circle:
    def __init__(self, radius):
        self._radius = radius #Private attribute
    
    @property
    def radius(self):
        #Getter method
        return self._radius
    
    @radius.setter
    def radius(self, radius):
        #Setter method using @property_name.setter decorator
        if radius < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = radius
circle = Circle(5)
print(circle.radius)
circle.radius = 10 #Set the radius using setter
print(circle.radius)
```

```python
#multiple class Inheritance

class A(object):
    def __init__(self):
        self.a = "Ram"
        print(self.a)

class B(object):
    def __init__(self):
        self.b = "Laxman"
        print(self.b)

class C(A, B):
    def __init__(self):
        self.C = "Hanuman"
        print(self.C)
        super().__init__() #Calls __init__ from class A

#Create an object of class C
o = C()
```

```python
#Method Resolution Order(MRO)
class A(object):
    def __init__(self):
        self.a = "Ram"
        print(self.a)
        super().__init__()
class B(object):
    def __init__(self):
        self.b = "Laxman"
        print(self.b)
        super().__init__()

class C(A, B):
    def __init__(self):
        self.C = "Hanuman"
        print(self.C)
        super().__init__()
o = C()
```

```python
#private attibutes in python
class PrivateClass:
    def __init__(self):
        self.public_data = "public" #public attribute
        self.__private_data = "private" #private attribute
#Creating an object of the class
my_object = PrivateClass()

#Accessing the public attribute
print(my_object.public_data) #This works fine

#Accing the private attribute directly will raise an error
try:
    print(my_object.__private_data) # This raises an AttributeError

except AttributeError as e:
    print(e)

#Accessing the private attribute using name managing
#print(my_object._PrivateClass.__private_data)
```

```python
#Unit testing with docstrings and Doctest
#Account class definition
from decimal import Decimal
class Account:
    #Account class for demonstrating dectest
    def __init__(self, account_number, balance):
        #Initialize an account object
        self.account_number = account_number
        self.balance = Decimal(balance)

    def deposit(self, amount):
        #Deposit money into the account
        self.balance += Decimal(amount)
        return self.balance
    
    def withdraw(self, amount):
        #Withdraw money from the account
        if self.balance >= Decimal(amount):
            self.balance -= Decimal(amount)
            return self.balance
        else:
            return "Insufficient funds"

#Unit test using doctest
```
