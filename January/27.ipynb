```python
class Student:
    rgno = 23
    name = "Rahul"
    branch = "CSE"
    def read(self):
        print("Reading")
    def write(self):
        print("Writing")

s1 = Student()
print("Student regno. : ", s1.rgno)
print("Student name : ", s1.name)
print("Student branch : ", s1.branch)
s1.read()
s1.write()
```

```python
class Student:
    def __init__(self, id, name, branch):
        self.id = id
        self.name = name
        self.branch = branch
    def display(self):
        print("Student id:", self.id)
        print("Student name:", self.name)
        print("Student branch:", self.branch)

s1 = Student(23, "Rahul", "CSE")
s1.display()
s2 = Student(3, "Retam", "CSE")
s2.display()
```
