```python
class Car:
    def __init__(self, model, color, speed):
        self.model = model
        self.color = color
        self.speed = speed
    def start(self):
        print(f"The {self.color} {self.model} car starts moving.")

class Audi(Car):
    def __init__(self, model, color, speed, year):
        super().__init__(model, color, speed)
        self.year = year

Audii7 =Audi(4,"white",100,2020)
print(Audii7.color)
Audii7.start()
```
