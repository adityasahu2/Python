```python
# Q1.
import random

# a)
numbers = [random.randint(1,101) for i in range(10)]
print (numbers)

# b)
# i)
sorted = sorted(numbers)
print(f"Second largest : {sorted[-2]}")
print(f"Second smallest : {sorted[1]}")

# ii)
even = [num for num in numbers if num % 2 == 0]
odd = [num for num in numbers if num % 2 == 1]

# iii)
even.sort(reverse=True)
odd.sort(reverse=True)
print(f"Sorted even : {even}")
print(f"Sorted odd : {odd}")
```

```python
# Q2.

# a)
str = input("Enter a string : ").strip()
words = str.split()
print(f"Number of words : {len(words)}")

# b)
frequency = {}
for char in str:
    if char != " ":
        if char not in frequency:
            frequency[char] = 1
        else:
            frequency[char] += 1
print(f"Character frequency : {frequency}")

# c)
max = max(frequency.values())
print(f"Most frequent character : {[key for key, value in frequency.items() if value == max]}")
```
