```python
from dataclasses import dataclass
@dataclass
class Person:
    name: str
    age: int
person = Person(name="Aditya", age=20)
print(person)
```

```python
import matplotlib.pyplot as plt
years = [2015,2016,2017,2018,2019,2020]
temperature = [10.5, 11.2, 12.8, 13.6, 14.1, 15.2]
plt.plot(years, temperature, label="Average Temperature", color="blue", marker="o")
plt.title("Average Temperature Over Years")
plt.xlabel("Years")
plt.ylabel("Temperature(c)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
years = [2015,2016,2017,2018,2019,2020]
temperature = [10.5, 11.2, 12.8, 13.6, 14.1, 15.2]
rainfall = [120, 100, 110, 140, 90, 110]
plt.plot(years, temperature, label="Average Temperature", color="blue", marker="o")
plt.plot(years, rainfall, label="Average Rainfall", color="green", marker="s")
plt.title("Average Temperature and Rainfall Over Years")
plt.xlabel("Years")
plt.ylabel("Values")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [2, 4, 6, 8]
y1 = [4, 16, 36, 34]
y2 = [8, 64, 214, 512]
y3 = [4, 8, 12, 16]
plt.plot(x, y1, label="y1", color="blue", marker="o")
plt.plot(x, y2, label="y2", color="green", marker="s")
plt.plot(x, y3, label="y3", color="red", marker="^")
plt.title("Line Graph")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
years = [2015,2016,2017,2018,2019,2020]
temperature = [10.5, 11.2, 12.8, 13.6, 14.1, 15.2]
plt.bar(years, temperature, label="Average Temperature", color="yellow", width=0.5)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.title("Customised Bar")
plt.xlabel("Years")
plt.ylabel("Temperature(c)")
plt.grid(True)
plt.legend()
plt.show()
```
