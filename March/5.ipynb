```python
def factorial(num):
    if num <= 1:
        return 1
    return num*factorial(num-1)
factorial(5)
```

```python
def fibonacci(num):
    if num in (0,1):
        return num
    else:
        return fibonacci(num-1) + fibonacci(num-2)
fibonacci(5)
```

```python
def linear_search(data, search):
    for index, value in enumerate(data):
        if value == search:
            return index
    return -1
```

```python
def binary_search(data, search):
    low = 0
    high = len(data) - 1
    while low <= high:
        mid = (low + high) // 2
        if data[mid] == search:
            return mid
        elif data[mid] < search:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```
